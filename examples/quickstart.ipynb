Run in an environment with this repository installed. Data retrieval requires internet. This notebook contains no precomputed observations. Confirm target and product availability first.

```python
from lc_tess.cli import main
TIC = 422526868  # example identifier; replace with your target
```

## Inspect archive products

This saves product inventories but does not download and analyze light curves.

```python
status = main(["--tic", str(TIC), "--search-only", "--output", "results_inventory"])
print("Exit status:", status)
```

## Retrieve and analyze

USC and SC remain separate. Additional clipping and flattening are off by default. Only run after reviewing the product inventory.

```python
status = main(["--tic", str(TIC), "--cadence", "both", "--output", "results"])
print("Exit status:", status)
```

## Read provenance and inspect outputs

A run can partly succeed. Inspect errors even when figures exist.

```python
from pathlib import Path
import json
folder = Path("results") / f"TIC_{TIC}"
manifest = json.loads((folder / "manifest.json").read_text())
manifest["errors"]
```

```python
from IPython.display import Image, display
for figure in sorted(folder.glob("*_diagnostics.png")):
    display(Image(filename=str(figure)))
```

## Optional localization

Install the localization extra first. Replace this empty list only with verified frequencies (microhertz) believed to come from the same source, and replace the sector. No frequency is automatically declared significant here.

```python
selected_frequencies_uhz = []
sector = 42  # example; replace after reviewing available observations
if selected_frequencies_uhz:
    status = main(["--tic", str(TIC), "--cadence", "sc", "--sectors", str(sector),
                   "--output", "results_localize", "--frequencies",
                   *map(str, selected_frequencies_uhz)])
```
